An MDP is five parts, and value iteration is one line applied until nothing moves. This notebook runs the three-state chain from the lesson and plots each sweep, so you can watch the value of `A` switch from the scenic route to the shortcut.

## 1. Value iteration on a chain

State A has two actions: a shortcut to B worth 0, and a scenic route to C worth 0.5. B leads to terminal C with reward 1. With `gamma = 0.9`, the shortcut should win.

```python
import numpy as np                              # array library, preinstalled on Colab

gamma = 0.9                                     # discount per step
V = np.zeros(3)                                 # A, B, C(terminal)
history = []                                    # one row per sweep

for sweep in range(6):                          # a few sweeps is enough here
    V[0] = max(0.0 + gamma * V[1],              # A: shortcut to B
               0.5 + gamma * V[2])              # A: scenic to terminal C
    V[1] = 1.0 + gamma * V[2]                   # B: go, reward 1
    V[2] = 0.0                                  # terminal stays zero
    history.append(V.copy())                    # record the sweep
    print(sweep, V.round(3))                    # 0.5 1 0, then 0.9 1 0
```

## 2. The sweeps as a picture

The first sweep picks the scenic route because B is still worth 0. Once B's value arrives, the shortcut overtakes it. The curves flatten when the Bellman equation reproduces itself.

```python
import matplotlib.pyplot as plt                 # plotting library, preinstalled on Colab

H = np.array(history)                           # shape: sweeps x states
plt.figure(figsize=(6, 3))                      # wide, short figure
plt.plot(H[:, 0], marker="o", label="V(A)")     # the choice state
plt.plot(H[:, 1], marker="s", label="V(B)")     # the reward state
plt.plot(H[:, 2], marker="^", label="V(C) terminal")
plt.xlabel("sweep")                             # x-axis
plt.ylabel("value")                             # y-axis
plt.legend()                                    # name the lines
plt.title("values settle at 0.9 and 1.0")        # takeaway
plt.show()                                      # render
```
