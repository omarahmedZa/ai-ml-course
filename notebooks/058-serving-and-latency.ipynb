Averages hide the requests users remember. In this notebook you compute nearest-rank percentiles by hand, then plot the tail that the mean leaves out.

## 1. Percentiles by hand

Twenty sorted request times. The nearest-rank rule is `rank = ceil(q / 100 * n)`, and the answer is the value at that 1-based position.

```python
import numpy as np                                    # arrays, preinstalled on Colab

lat = np.array([30, 32, 33, 35, 36, 38, 40, 41, 42, 45,
                47, 50, 52, 55, 60, 70, 85, 110, 150, 400])   # milliseconds
lat_sorted = np.sort(lat)                             # ascending order
for q in (50, 95, 99):                                # three percentiles
    rank = int(np.ceil(q / 100 * len(lat)))           # 1-based position
    print(f"p{q} rank={rank:2d} value={lat_sorted[rank - 1]:3d} ms")
print(f"mean = {lat.sum() / len(lat):.2f} ms")        # 72.55, below p95
```

## 2. Draw the tail

Plot each request against its position in the sorted list. The left side is the calm middle; the right edge is what times out.

```python
import matplotlib.pyplot as plt                       # plotting, preinstalled on Colab

positions = np.arange(1, len(lat_sorted) + 1)         # 1..20
plt.figure(figsize=(6, 3))                            # wide, short figure
plt.plot(positions, lat_sorted, "o-", color="tab:blue")   # sorted latency curve
plt.axhline(lat_sorted[9], color="grey", linestyle="--")   # p50 = 45
plt.axhline(lat_sorted[18], color="orange", linestyle="--")  # p95 = 150
plt.axhline(lat_sorted[19], color="red", linestyle="--")   # p99 = 400
plt.xlabel("position in sorted list")                  # rank, not request id
plt.ylabel("latency (ms)")                            # milliseconds
plt.title("the last point sets p99")                   # takeaway
plt.show()                                            # render
```
