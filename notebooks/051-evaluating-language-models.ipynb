A language model gives a probability to every token. This notebook turns those probabilities into perplexity, then shows why a benchmark score needs an error bar before you compare two models. Everything runs on a fresh Colab runtime.

## 1. Perplexity, one token at a time

Perplexity is `exp` of the average negative log-probability. Start with four tokens whose probabilities are 1.0, 0.5, 0.5 and 0.25 — one certain token and three uncertain ones.

```python
import numpy as np                              # array library, preinstalled on Colab

p = np.array([1.0, 0.5, 0.5, 0.25])             # probability the model gave each true token
log_p = np.log(p)                               # natural log: 0, -0.693, -0.693, -1.386
nll = -log_p.mean()                             # average surprise = 0.693
ppl = np.exp(nll)                               # push back through exp

print("log p:", log_p.round(3))                 # [ 0.    -0.693 -0.693 -1.386]
print("mean nll:", round(float(nll), 3))        # 0.693
print("perplexity:", round(float(ppl), 3))      # 2.0
```

## 2. Why accuracy needs an error bar

A benchmark score is a proportion. Its standard error is `sqrt(p(1-p)/n)`, so the same 70% is precise on 2,000 questions and fuzzy on 50. Plot the half-width against sample size.

```python
import matplotlib.pyplot as plt                 # plotting library, preinstalled on Colab

sizes = np.array([10, 20, 50, 100, 200, 500, 1000, 2000])   # questions per benchmark
p_hat = 0.70                                    # the observed accuracy
se = np.sqrt(p_hat * (1 - p_hat) / sizes)       # standard error at each size

for n, s in zip(sizes, se):                     # print the table
    print(f"n={n:5d}  se={s:.3f}  95% +/- {2*s:.3f}")

plt.figure(figsize=(6, 3))                      # wide, short figure
plt.plot(sizes, 2 * se, marker="o", color="tab:red")   # 95% half-width
plt.xscale("log")                               # sample size spans two decades
plt.xlabel("questions")                         # x-axis
plt.ylabel("95% half-width")                    # y-axis
plt.title("four times the questions halves the error bar")   # takeaway
plt.show()                                      # render
```
