A convolution slides one small filter across an image, taking a dot product at every position. In this notebook you compute a vertical-edge response by hand, plot the feature map, and check the output-size formula.

```python
import numpy as np                                        # the array library
img = np.zeros((5, 5))                                    # a blank grey image
img[:, 2:] = 1.0                                          # brighten the right-hand side
k = np.array([[-1., 0., 1.], [-1., 0., 1.], [-1., 0., 1.]])  # vertical edge filter
out = np.zeros((3, 3))                                    # output map, no padding
for i in range(3):                                        # three output rows
    for j in range(3):                                    # three output columns
        patch = img[i:i + 3, j:j + 3]                     # the patch under the filter
        out[i, j] = (patch * k).sum()                     # elementwise product, then sum
print(out)                                                # rows of 3, 3, 0
side = (5 + 2 * 0 - 3) // 1 + 1                           # output-size formula
print("formula says", side, "array says", out.shape[0])   # both should be 3
```

## What the numbers say

The filter subtracts the left column from the right. A window straddling the edge scores 3; a window entirely inside the bright region scores 0. The whole map is three identical rows of `3, 3, 0`, because the edge is vertical and runs the full height.

```python
import matplotlib.pyplot as plt                           # plotting library
fig, axes = plt.subplots(1, 2, figsize=(7, 3))            # image and feature map side by side
axes[0].imshow(img, cmap="gray", vmin=0, vmax=1)          # the input image
axes[0].set_title("input 5x5")                             # label the input
axes[1].imshow(out, cmap="RdBu")                          # the convolution output
axes[1].set_title("feature map 3x3")                       # label the output
for ax in axes:                                           # both panels
    ax.axis("off")                                        # hide the tick marks
plt.suptitle("The filter responds where the edge is")     # the takeaway
plt.show()                                                # render the figure
```

Change the image so the bright region is a single column at index 2: `img[:] = 0; img[:, 2] = 1`. Predict the feature map before you rerun. Then set `stride = 2` in a loop rewrite and confirm the map shrinks to 2x2, as `(5 - 3)//2 + 1 = 2` says it should.

```python
def conv_params(c_in, c_out, k):                          # parameters for one conv layer
    return c_out * (c_in * k * k + 1)                     # filters times weights plus bias
def out_side(size, k, stride=1, pad=0):                   # output side of a convolution
    return (size + 2 * pad - k) // stride + 1             # the sizing formula
print("3x3, 3 in, 64 out:", conv_params(3, 64, 3))        # 1792
print("224 with k=3, s=2, p=1:", out_side(224, 3, 2, 1))  # 112
print("32 with k=5, s=1, p=2:", out_side(32, 5, 1, 2))    # 32
dense = 224 * 224 * 3 * 64                                # dense weights for 64 outputs
print("dense equivalent:", dense)                         # 9633792
print("ratio:", round(dense / conv_params(3, 64, 3)))     # about 5376x
```
