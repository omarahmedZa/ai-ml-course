With 2% positives, a model that never predicts the positive class is 98% accurate. This notebook builds a confusion matrix from counts, computes precision, recall and F1, and shows why accuracy alone hides the failure.

```python
import numpy as np                                        # the array library
tp, fp, fn, tn = 10, 90, 10, 890                          # counts from the worked example
precision = tp / (tp + fp)                                # of the alerts raised, how many real
recall = tp / (tp + fn)                                   # of the real positives, how many found
f1 = 2 * precision * recall / (precision + recall)        # harmonic mean of the two
accuracy = (tp + tn) / (tp + fp + fn + tn)                # share of all rows correct
print("accuracy", accuracy)                              # 0.9
print("precision", precision)                            # 0.1
print("recall", recall)                                  # 0.5
print("f1", round(f1, 3))                                # 0.167
```

## What the numbers say

Ninety percent accuracy sounds strong until the precision line: only one alert in ten is real. The model finds half the fraud and sends 90 false alarms to review. Every metric here is computed from the same four counts, so the confusion matrix is the object to inspect first.

```python
import matplotlib.pyplot as plt                           # plotting library
names = ["accuracy", "precision", "recall", "F1"]        # the four metrics
values = [accuracy, precision, recall, f1]                # their values
colors = ["grey", "tab:blue", "tab:orange", "tab:red"]  # one colour each
plt.bar(names, values, color=colors)                      # draw the comparison
plt.ylim(0, 1)                                            # all metrics live in 0..1
plt.ylabel("score")                                       # y-axis label
plt.title("98% accuracy can hide a 0.17 F1")              # the takeaway
plt.show()                                                # render the figure
```

## Your turn

The grey bar towers over the rest. Now price the baseline that predicts "negative" for every row: accuracy becomes 0.98, recall becomes 0, and precision is undefined because no alerts are raised. Change the counts to `tp, fp, fn, tn = 0, 0, 20, 980` and rerun to see the arithmetic refuse to divide by zero.

```python
global_mean = 0.2                                         # base rate of the positive class
hits, rows, strength = 1, 1, 10                           # one category, one hit, m = 10
smoothed = (hits + strength * global_mean) / (rows + strength)  # blend with the prior
print("smoothed encoding", round(smoothed, 3))            # 0.273
big_hits, big_rows = 30, 100                              # a category with 100 rows
big = (big_hits + strength * global_mean) / (big_rows + strength)  # less shrinkage
print("well-populated category", round(big, 3))           # 0.291
print("raw encoding was", 1.0)                           # the leak: the label itself
```
