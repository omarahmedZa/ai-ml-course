A support vector machine picks the separating line with the widest empty strip. In this notebook you compute that margin, price mistakes with hinge loss, and build an RBF kernel matrix by hand. Everything runs on a fresh Colab runtime with NumPy and Matplotlib.

```python
import numpy as np                                        # the array library
w = np.array([3., 4.])                                    # one weight per feature
margin = 2 / np.linalg.norm(w)                            # margin = 2 / length of w
print("margin =", margin)                                 # 0.4
scores = np.array([1.0, 0.5, -0.5])                       # three model scores
labels = np.array([1, 1, -1])                             # +1 or -1 per row
hinge = np.maximum(0, 1 - labels * scores)                # per-row hinge loss
print("mean hinge =", round(float(hinge.mean()), 3))       # 0.333
X = np.array([[1., 0.], [0., 1.], [2., 2.]])              # three points, two features
d2 = ((X[:, None, :] - X[None, :, :]) ** 2).sum(-1)       # squared distances
K = np.exp(-1.0 * d2)                                     # RBF kernel, gamma = 1
print(np.round(K, 3))                                     # similarity matrix
```

## What the numbers say

`w = [3, 4]` has length 5, so the margin is `2/5 = 0.4`. The first row scores exactly 1, which is on the margin and costs nothing; the third row is wrong and costs 1.5. The kernel matrix is symmetric with ones on the diagonal: every point is perfectly similar to itself.

```python
import matplotlib.pyplot as plt                           # plotting library
s = np.linspace(-2, 3, 120)                               # scores from -2 to 3
loss = np.maximum(0, 1 - 1 * s)                           # hinge loss for y = +1
plt.plot(s, loss, color="tab:red")                        # draw the hinge curve
plt.axvline(1, color="grey", linestyle="--")              # score of 1 costs zero
plt.xlabel("score")                                       # x-axis label
plt.ylabel("loss")                                        # y-axis label
plt.title("Hinge loss is flat once the margin is cleared") # the takeaway
plt.show()                                                # render the figure
```

## Your turn

The curve is flat to the right of 1 and rises to the left. That flat region is the margin doing its job: confident, correct points stop pulling on the boundary. Change the label to `-1` in the loss line and predict which way the kink moves before you rerun.

```python
d = np.linspace(0, 3, 120)                                # distances between points
for gamma in (0.5, 1.0, 4.0):                             # three similarity reaches
    sim = np.exp(-gamma * d ** 2)                         # RBF similarity at this gamma
    plt.plot(d, sim, label=f"gamma={gamma}")              # one curve per gamma
plt.xlabel("distance between points")                      # x-axis label
plt.ylabel("similarity")                                   # y-axis label
plt.legend()                                              # name the curves
plt.title("Large gamma: similarity dies within one unit")  # the takeaway
plt.show()                                                # render the figure
```
