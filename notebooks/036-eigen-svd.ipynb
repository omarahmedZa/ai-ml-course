A matrix turns most vectors. A few special directions only get stretched — those are the eigenvectors, and the stretch factors are the eigenvalues. This notebook checks that by hand, then draws the unit circle being stretched into an ellipse. Everything runs on a fresh Colab runtime with NumPy and Matplotlib only.

```python
import numpy as np  # arrays and linear algebra
A = np.array([[3., 1.], [1., 3.]])  # the symmetric matrix from the lesson
vals, vecs = np.linalg.eigh(A)  # eigenvalues and unit eigenvectors
print("eigenvalues", vals)  # [2. 4.]
print("vector for 2", vecs[:, 0])  # [0.707, -0.707]
print("vector for 4", vecs[:, 1])  # [0.707, 0.707]
print("A v", A @ vecs[:, 1])  # [2.83, 2.83]
print("4 v", 4 * vecs[:, 1])  # [2.83, 2.83]
print("check", np.allclose(A @ vecs[:, 1], 4 * vecs[:, 1]))  # True
```

## What just happened

`eigh` is the routine for symmetric matrices: it returns real eigenvalues sorted upward and orthonormal eigenvectors as columns. The second column is the dominant direction, and `A @ v` equals `4 * v` to machine precision.

### The SVD of the same matrix

Because `A` is symmetric and positive, its singular values are just its eigenvalues. For a general matrix they are the square roots of the eigenvalues of `A.T @ A`, and they are always non-negative.

```python
import matplotlib.pyplot as plt  # plotting
theta = np.linspace(0, 2 * np.pi, 200)  # angles around the unit circle
circle = np.vstack([np.cos(theta), np.sin(theta)])  # 2 x 200 points
ellipse = A @ circle  # map every point through the matrix
plt.figure(figsize=(5, 5))  # square figure so shapes stay honest
plt.plot(circle[0], circle[1], "--", color="grey", label="unit circle")  # input
plt.plot(ellipse[0], ellipse[1], color="tab:blue", label="A x circle")  # output
plt.axhline(0, color="grey", linewidth=0.6)  # horizontal axis
plt.axvline(0, color="grey", linewidth=0.6)  # vertical axis
plt.gca().set_aspect("equal")  # equal units on both axes
plt.legend()  # name both curves
plt.title("the matrix stretches the circle into an ellipse")  # takeaway
plt.show()  # draw it
```

## Reading the picture

The dashed circle is every vector of length 1. The solid curve is where those vectors land. The longest radius is the largest singular value, 4, and the shortest is 2. The two radii are perpendicular because the matrix is symmetric, and they point along the eigenvectors.

The SVD writes the same facts for any matrix: `A = U @ diag(S) @ Vt`. `U` holds output directions, `Vt` holds input directions, and `S` holds the stretch factors in descending order.

```python
def power_iteration(M, steps=20):  # find the top direction without eigh
    v = np.ones(M.shape[0])  # start along the all-ones vector
    for _ in range(steps):  # repeat the multiplication
        w = M @ v  # apply the matrix
        v = w / np.linalg.norm(w)  # renormalise to unit length
    return v  # converged direction
v = power_iteration(np.array([[3., 1.], [1., 3.]]))  # run it
print("direction", v)  # about [0.707, 0.707]
print("Rayleigh", v @ np.array([[3., 1.], [1., 3.]]) @ v)  # about 4.0
```
