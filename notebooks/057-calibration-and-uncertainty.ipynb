A probability is a promise about frequency. Here you bin predictions, compute expected calibration error from the lesson's table, and draw the reliability diagram that shows where the promise breaks.

## 1. Expected calibration error, bin by bin

The five bins below come from 200 held-out predictions. Each weight is the bin's share of the data; ECE is the weighted sum of the confidence gaps.

```python
import numpy as np                                    # arrays, preinstalled on Colab

conf = np.array([0.15, 0.30, 0.50, 0.70, 0.90])       # mean prediction per bin
obs = np.array([0.10, 0.25, 0.52, 0.55, 0.75])        # observed positive rate
n = np.array([20, 30, 50, 60, 40])                    # rows per bin
weights = n / n.sum()                                 # shares, sum to 1
gaps = np.abs(conf - obs)                             # one gap per bin
contrib = weights * gaps                              # weighted gaps
ece = contrib.sum()                                   # expected calibration error
for i in range(len(n)):                               # print the working
    print(f"bin {i}: weight={weights[i]:.2f} gap={gaps[i]:.2f} contribution={contrib[i]:.4f}")
print("ECE =", round(ece, 4))                         # expect 0.0925
```

## 2. The reliability diagram

Perfect calibration is the diagonal. Points below it are overconfident: the model claimed more certainty than the observed frequency supports.

```python
import matplotlib.pyplot as plt                       # plotting, preinstalled on Colab

plt.figure(figsize=(5, 4))                            # square-ish reliability plot
plt.plot([0, 1], [0, 1], "--", color="grey", label="perfect")   # y = x
plt.plot(conf, obs, "o-", color="tab:red", label="model")       # the five bins
plt.xlabel("predicted confidence")                    # x-axis
plt.ylabel("observed frequency")                      # y-axis
plt.title("above the line = underconfident")           # short takeaway
plt.legend(loc="upper left")                          # name both series
plt.xlim(0, 1); plt.ylim(0, 1)                        # keep the diagonal square
plt.show()                                            # render
```
